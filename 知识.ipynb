数据库级别的长期记忆，核心就是这条链路：

用户消息
   ↓
MemoryExtractor：提取值得长期保存的信息
   ↓
MemoryWriter：转换成统一结构
   ↓
MemoryRepository：写入 PostgreSQL
   ↓
PostgreSQL 长期保存


下一次用户发起任务时：

user_id
   ↓
从 PostgreSQL 查询该用户的历史记忆
   ↓
MemoryContext：转换成文本
   ↓
注入 Agent 的 Prompt
   ↓
Agent 根据历史偏好做决策
